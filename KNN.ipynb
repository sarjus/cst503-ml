```python
#This example demonstrates binary classification using a small dataset.
#Step 1: Import Required Libraries
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
```

```python
'''
Step 2: Create the Training Dataset

Features:

Study Hours
Attendance

Target:

0 = Fail
1 = Pass
'''
# Training data
X_train = np.array([
    [2, 60],
    [3, 70],
    [5, 75],
    [6, 90],
    [7, 85]
])

# Class labels
y_train = np.array([0, 0, 1, 1, 1])
```

```python
#Step 3: Create the KNN Model
# Create KNN model
knn = KNeighborsClassifier(
    n_neighbors=3,
    metric='euclidean'
)
```

```python
#Step 4: Train the Model
# Train the model
knn.fit(X_train, y_train)
```

```python
'''
Step 5: Predict a New Student

New Student:

Study Hours = 4
Attendance = 80
'''
# Test sample
X_test = np.array([[4, 80]])

# Prediction
prediction = knn.predict(X_test)
```

```python
#Step 6: Display the Result
if prediction[0] == 1:
    print("Prediction: Pass")
else:
    print("Prediction: Fail")
```
